- `"contagem"`: quantos salários foram submetidos
- `"media"`: média dos salários, arredondada para 2 casas decimais
- `"minimo"`: menor valor

**Exemplo:**

````python
# sua solução aqui

salarios = [50000, 80000, 120000, 95000, 60000]
def resumo_salarios(salarios):
  s = np.array(salarios)
  res = {
        "contagem": len(salarios),
        "media": round(s.mean(), 2),
        "minimo": s.min(),
        "maximo": s.max()
    }
  return res

resultado = resumo_salarios(salarios)

print(resultado)
````

### Parte A — Representar um candidato

**Função 1:** `criar_candidato(nome, cargo, anos, pretensao, modalidade)`

Retorna um dicionário no formato descrito acima.

**Exemplo:**


**Função 2:** `apresentar_candidato(candidato)`

Recebe um dicionário de candidato e retorna uma string formatada.

**Exemplo de saída:**

```
Ana Lima
  Cargo desejado:  Analista de Dados
  Nível:           Júnior (1 ano de experiência)
  Pretensão:       USD 55,000/ano
  Modalidade:      remoto
```

**Função 3:** `esta_na_faixa(candidato, salario_min, salario_max)`

Retorna `True` se a pretensão salarial estiver no intervalo (inclusive nos extremos), e `False` caso contrário.

**Exemplo:**

```
ana = criar_candidato("Ana Lima", "Analista", 1, 55000, "remoto")
esta_na_faixa(ana, 50000, 60000)  ->  True
esta_na_faixa(ana, 60000, 80000)  ->  False
```

````python
  # sua solução aqui
def criar_candidato(nome, cargo, anos, pretensao, modalidade):

      candidato = {
        "nome": nome,
        "cargo_desejado": cargo,
        "anos_experiencia": anos,
        "pretensao_salarial": pretensao,
        "modalidade": modalidade
        }

      return candidato


# sua solução aqui
def apresentar_candidato(candidato):

    anos = candidato["anos_experiencia"]
    if anos <= 2:
      nivel = "Júnior"
    elif anos <= 5:
      nivel = "Pleno"
    else:
      nivel = "Sênior"

    texto = f"""
  {candidato["nome"]}
    Cargo Desejado:   {candidato["cargo_desejado"]}
    Nível:            {nivel} ({anos} ano(s) de experiência)
    Pretensão:        USD {candidato["pretensao_salarial"]:,}/ano
    Modalidade:       {candidato["modalidade"]}
  """
    return texto


# sua solução aqui
def esta_na_faixa(candidato, salario_min, salario_max):

    pretensao = candidato["pretensao_salarial"]

    return salario_min <= pretensao <= salario_max

````

**Função 9:** `ranking_por_experiencia(banco)`

Retorna a lista de candidatos ordenada do mais experiente para o menos experiente.

**Exemplo:**

```
ranking = ranking_por_experiencia(banco)
ranking[0]["nome"]   ->  "Carla Mendes"  (8 anos)

```

````python
# sua solução aqui
def media_pretensao(banco):

    soma = 0

    for candidato in banco:
        soma += candidato["pretensao_salarial"]

    media = soma / len(banco)

    return media


# sua solução aqui
def candidatos_na_faixa(banco, salario_min, salario_max):
    resultado = []
    for candidato in banco:
         pretensao = candidato["pretensao_salarial"]
         if salario_min <= pretensao <= salario_max:
             resultado.append(candidato)
    return resultado


# sua solução aqui
def ranking_por_experiencia(banco):

    ranking = sorted(
        banco,
        key=lambda candidato: candidato["anos_experiencia"],
        reverse=True
    )

    return ranking
````

## **Desafio bônus** — Validação com exceções

Escreva uma função `validar_dados_candidato(dados)` que recebe um dicionário e levanta uma exceção `ValueError` com mensagem clara em cada um destes casos:

- Falta alguma das chaves obrigatórias: `nome`, `cargo_desejado`, `anos_experiencia`, `pretensao_salarial`, `modalidade`
- `anos_experiencia` é negativo
- `pretensao_salarial` é negativa
- `modalidade` não é `"remoto"`, `"presencial"` ou `"híbrido"`

Se todos os dados forem válidos, a função retorna `True`.

**Exemplo:**

```
validar_dados_candidato({"nome": "Ana", ...})       ->  True
validar_dados_candidato({"nome": "Ana"})            ->  ValueError: chave faltando: cargo_desejado
validar_dados_candidato({..., "modalidade": "casa"}) ->  ValueError: modalidade inválida: casa
```

````python
# sua solução aqui
def validar_dados_candidato(dados):
  chaves_obrigatorias = [
      "nome",
      "cargo_desejado",
      "anos_experiencia",
      "pretensao_salarial",
      "modalidade"
      ]

  for chave in chaves_obrigatorias:
    if chave not in dados:
      raise ValueError(f"chave faltando: {chave}")

  if dados["anos_experiencia"] < 0:
    raise ValueError("anos_experiencia é negativo")

  if dados["pretensao_salarial"] < 0:
    raise ValueError("pretensao_salarial é negativo")

    validas = ["remoto", "presencial", "híbrido"]

    if dados["modalidade"] not in validas:
      raise ValueError(f"modalidade inválida: {dados['modalidade']}")

  return True


````

````python
# Testes do bônus
dados_validos = {
    "nome": "Ana Lima",
    "cargo_desejado": "Analista de Dados",
    "anos_experiencia": 1,
    "pretensao_salarial": 55000,
    "modalidade": "remoto"
}

print(validar_dados_candidato(dados_validos) == True)

# Testes de erro — devem levantar ValueError
casos_invalidos = [
    {"nome": "Ana"},
    {**dados_validos, "anos_experiencia": -1},
    {**dados_validos, "pretensao_salarial": -100},
    {**dados_validos, "modalidade": "casa"},
]

for caso in casos_invalidos:
    try:
        validar_dados_candidato(caso)
        print(False, "deveria ter levantado ValueError:", caso)
    except ValueError as e:
        print(True, "ValueError:", e)
````
